# OmniVoice Generation

Generate speech from a checkpoint, save it as a WAV file, and report generation speed.

```python
from pathlib import Path
import sys
import time

import soundfile as sf
import torch
import numpy as np
from IPython.display import Audio

from omnivoice.models.omnivoice import OmniVoice
```

```python
# Generation configuration.
import os
checkpoint = '../ner01/finetune/exp/omnivoice_finetune/checkpoint-2000'
# checkpoint = os.path.abspath(checkpoint)

language = 'lt'
device = "cpu"

num_step = 32
guidance = 4.5

print(f'Checkpoint: {checkpoint}')
print(f'Device: {device}')
print(f'num_step: {num_step}')
print(f'guidance: {guidance}')
print('Loading checkpoint...')
model = OmniVoice.from_pretrained(
    checkpoint,
    device_map=device,
    dtype=torch.float16,
)

def tts(model, text):
    started = time.perf_counter()
    with torch.inference_mode():
        audio = model.generate(
            text=text,
            language=language,
            num_step=num_step,
            guidance_scale=guidance,
            class_temperature=0.0,
            position_temperature=5.0,
        )[0]
    generation_seconds = time.perf_counter() - started
    audio_duration = len(audio) / model.sampling_rate
    print(f'  Duration: {audio_duration:.2f}s')
    print(f'  In: {generation_seconds:.2f}s')
    print(f'  RTF: {generation_seconds / audio_duration:.2f}x')
    return np.asarray(audio, np.float32)

print(f'Loaded at {model.sampling_rate} Hz')
```
